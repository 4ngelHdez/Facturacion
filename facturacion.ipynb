## Agrupando datos

```python
# Se obtiene la suma de la columna 'precio' agrupada por la columna 'tienda'.
# Esto nos permite ver el total de ventas por tienda.

data.groupby('tienda')['precio'].sum().to_frame()
```

```python
# Se obtiene la suma de la columna 'precio' agrupada por la columna 'forma_pago'.
# Esto nos permite ver el total de ventas por forma de pago.

data.groupby('forma_pago')['precio'].sum().to_frame()
```

```python
# Se obtiene la suma de la columna 'precio' agrupada por las columnas 'tienda' y 'forma_pago'.
# Esto nos permite ver el total de ventas de cada tienda en sus diferentes formas de pago.

data_agrupada = data.groupby(['tienda', 'forma_pago'])['precio'].sum().to_frame()
data_agrupada
```

```python
# Se exporta la data agrupada a un archivo Excel.

data_agrupada.to_excel("facturación.xlsx")
```

## Visualización

```python
# Elaboración de un grafico de barras para visualizar la facturación por cada columna de interés.

lista_columnas = ['tienda', 'ciudad', 'pais', 'tamano', 'local_consumo']

for columna in lista_columnas:
    fig = px.histogram(data, x=columna,
             y = 'precio',
             text_auto = True,
             title = "Facturación",
             color = "forma_pago")

    fig.show()
    fig.write_html(f'Facturación-{columna}.html')
```

## Conclusiones 

* La ciudad que mas facturas emitio fue Guadalajara con 13483 operaciones, seguida de Santiago de Chile con 13075, Buenos Aires quedo en tercer lugar con 12344, Ciudad de México en cuarto con 12177, Bogotá y Medellín quedaron en quinto y sexto respectivamente con 10603 y 8318 operaciones.

* Se cuentan con 4 métodos de pago: Crédito, Débito, Efectivo y Nequi.

* En todos los gráficos que hicimos por las columnas de interés se muestra la misma tendencia en cuanto a la preferencia de los usuarios para el pago de sus compras ya que el pagar con tarjeta es su preferida, en primera instancia se inclinan en pagar con la tardeta de crédito, muy de cerca están los que prefieren pagar con tarjeta de debido, en tercer lugar están los amantes del dinero corriente y por último esta nueva tecnología de pago como lo es Nequi.
